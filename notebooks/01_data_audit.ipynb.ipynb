## 5. Audit de qualité des données

L'audit mesure, pour chaque variable, son type, le nombre de valeurs renseignées, le nombre et le pourcentage de valeurs manquantes, ainsi que le nombre de valeurs distinctes.

```python
audit_raw = pd.DataFrame({
    "Type de données": df.dtypes.astype(str),
    "Valeurs non nulles": df.notna().sum(),
    "Valeurs manquantes": df.isna().sum(),
    "Pourcentage manquant (%)": (df.isna().mean() * 100).round(2),
    "Valeurs distinctes": df.nunique(dropna=True)
})

audit_raw = audit_raw.sort_values(
    by=["Valeurs manquantes", "Valeurs distinctes"],
    ascending=[False, True]
)

display(audit_raw)
```

## 6. Dictionnaire de données et règles métier

Le dictionnaire ci-dessous est généré à partir des colonnes réellement présentes dans le fichier.

La liste des variables dites **potentiellement disponibles à l'inscription** est une hypothèse de travail. Elle doit être validée avec la documentation du jeu de données et le contexte métier avant toute modélisation.

```python
# Variables considérées comme potentiellement disponibles au moment de l'inscription.
# Cette liste constitue une hypothèse métier à valider.
variables_inscription_potentielles = [
    "Marital status",
    "Application mode",
    "Application order",
    "Course",
    "Daytime/evening attendance",
    "Previous qualification",
    "Nacionality",
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    "Admission grade",
    "Displaced",
    "Educational special needs",
    "Debtor",
    "Tuition fees up to date",
    "Gender",
    "Scholarship holder",
    "Age at enrollment",
    "International"
]

dictionnaire_donnees = {}

for col in df.columns:
    if pd.api.types.is_bool_dtype(df[col]):
        type_metier = "Booléen"
    elif pd.api.types.is_numeric_dtype(df[col]):
        type_metier = "Numérique"
    elif pd.api.types.is_datetime64_any_dtype(df[col]):
        type_metier = "Date/Heure"
    else:
        type_metier = "Catégoriel / Texte"

    dictionnaire_donnees[col] = {
        "Description": f"Variable « {col} » présente dans le jeu de données ; description métier à confirmer.",
        "Type métier estimé": type_metier,
        "Potentiellement disponible à l'inscription": (
            "Oui" if col in variables_inscription_potentielles else "À vérifier"
        ),
        "Valeurs manquantes": int(df[col].isna().sum()),
        "Valeurs distinctes": int(df[col].nunique(dropna=True))
    }

dic_df = pd.DataFrame.from_dict(dictionnaire_donnees, orient="index")
dic_df.index.name = "Nom de la variable"

display(dic_df)
```

## 7. Variables potentiellement disponibles à l'inscription

Cette vue filtre les variables marquées « Oui » dans la liste de travail. Ce classement ne constitue pas une garantie de disponibilité réelle : il faut confirmer que les informations existent bien au moment où la prédiction sera effectuée.

```python
variables_inscription = dic_df[
    dic_df["Potentiellement disponible à l'inscription"] == "Oui"
].copy()

print(f"Nombre de variables retenues dans la liste de travail : {len(variables_inscription)}")
display(variables_inscription)
```

## 3. Interprétation du type métier

Le type informatique d'une variable ne correspond pas toujours à son type métier.

Par exemple, `Marital status` peut être stockée sous forme d'entier dans le fichier CSV. Cependant, les valeurs 1, 2, 3, 4, 5 et 6 représentent des catégories d'état matrimonial.

Il convient donc de distinguer :

* **Variable catégorielle nominale :** représente des catégories sans ordre naturel, comme la nationalité, la formation ou la profession.
* **Variable catégorielle ordinale :** représente des catégories qui peuvent être classées selon un ordre pertinent, comme certains niveaux de qualification.
* **Variable binaire :** représente deux modalités, souvent codées 0 et 1.
* **Variable quantitative :** représente une mesure ou un dénombrement, comme l'âge, les notes ou le nombre d'unités validées.

## 4. Précaution pour la modélisation

Le choix des variables doit tenir compte du moment où le modèle sera utilisé.

Si l'objectif est d'identifier les étudiants à risque dès leur inscription, les résultats du premier et du deuxième semestre ne sont pas encore disponibles et ne doivent pas être utilisés comme variables explicatives.

Si l'objectif est d'estimer le risque après un semestre, les informations disponibles à cette date peuvent être utilisées, à condition de définir précisément la cible et la période de prédiction.

Cette précaution permet de construire un modèle cohérent avec son utilisation réelle.
